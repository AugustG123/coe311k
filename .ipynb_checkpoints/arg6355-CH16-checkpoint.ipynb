```python
import numpy as np
import matplotlib.pyplot as plt

# Define the parameters
k = 3.0        # Constant in the equation (e.g., spring constant)
h1 = [.01, .02, .03, .05, .1, .2, .3, .5]      # Step size
x_start = 0    # Start of the interval
x_end = 10     # End of the interval
y_0 = 1.0      # Initial condition for y
v_0 = 0.0      # Initial condition for dy/dx

for h in h1:
# Create arrays to store x, y, and v values
    x_values = np.arange(x_start, x_end + h, h)
    y_values = np.zeros(len(x_values))
    v_values = np.zeros(len(x_values))

# Set initial conditions
    y_values[0] = y_0
    v_values[0] = v_0

# Euler's Method loop
    for i in range(1, len(x_values)):
    # Current values
        x_n = x_values[i - 1]
        y_n = y_values[i - 1]
        v_n = v_values[i - 1]    

    # Update equations
        y_values[i] = y_n + h * v_n
        v_values[i] = v_n + h * (-k * y_n)

# Plot the results
    plt.xlabel('x')
    plt.ylabel('y and v')
    plt.title('Solution of d²y/dx² = -ky using Euler\'s Method')
    plt.legend()
    plt.grid(True)
    plt.plot(x_values, y_values, label='h value of ' + str(h))     
plt.show()
```

The graph becomes extremely unstable at h =.5 with some instability at h=.3,.2. The larger the k, the greater the instability.
