```python
import numpy as np
import matplotlib.pyplot as plt
h = (2*np.pi - 0) / (50 - 1)
x = np.linspace(0,2*np.pi, 50)

# Central Difference Approximation for the derivative
def f(x):
    return np.sin(x)
df_dxC = (f(x + h) - f(x - h)) / (2 * h)

# Forward Difference Approximation
### YOUR CODE HERE ###

df_dxF = (f(x) - f(x-h))/h

# Backward Difference Approximation
df_dxB = (f(x+h) - f(x))/h


fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

#all curves
ax1.plot(x, np.cos(x), label='Exact: cos(x)', linewidth=2, color='black')
ax1.plot(x, df_dxF,   label='Forward',  linestyle='--', color='steelblue')
ax1.plot(x, df_dxB,   label='Backward', linestyle=':',  color='tomato')
ax1.plot(x, df_dxC,   label='Central',  linestyle='-.', color='seagreen')
ax1.set_title("Derivative Approximations")
ax1.set_xlabel("x")
ax1.set_ylabel("f'(x)")
ax1.legend()
ax1.grid(True)

#absolute error
ax2.plot(x, np.abs(df_dxF - np.cos(x)), label='Forward error',  linestyle='--', color='steelblue')
ax2.plot(x, np.abs(df_dxB - np.cos(x)), label='Backward error', linestyle=':',  color='tomato')
ax2.plot(x, np.abs(df_dxC - np.cos(x)), label='Central error',  linestyle='-.', color='seagreen')
ax2.set_title("Absolute Error vs Exact Derivative")
ax2.set_xlabel("x")
ax2.set_ylabel("|error|")
ax2.legend()
ax2.grid(True)

plt.suptitle("f(x) = sin(x)  —  Step size h = {:.4f}".format(h), fontsize=13)
plt.tight_layout()
plt.show()
```
