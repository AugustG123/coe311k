```python
import numpy as np
import matplotlib.pyplot as plt

def f(x, y):
    return -2 * y

x_exact = np.linspace(0, 2, 200)
y_exact = np.exp(-2 * x_exact)

plt.figure(figsize=(10, 6))

for h in np.linspace(0.01, 1, 20):
    x_start, x_end = 0, 2
    num_steps = int((x_end - x_start) / h)
    x_values = np.linspace(x_start, x_end, num_steps + 1)
    y_values = np.zeros(num_steps + 1)

    # Initial condition
    y_values[0] = 1

    # Euler's method
    for i in range(num_steps):
        y_values[i + 1] = y_values[i] + h * f(x_values[i], y_values[i])

    # Plot inside the loop, label each h
    plt.plot(x_values, y_values, label=f"h = {h:.2f}", alpha=0.7)

# Plot exact solution for reference
plt.plot(x_exact, y_exact, 'k--', linewidth=2, label="Exact: e^(-2x)")

plt.xlabel('x')
plt.ylabel('y')
plt.title("Approximation of dy/dx = -2y using Euler's Method")
plt.legend(fontsize=7, ncol=2)
plt.grid(True)
plt.show()
```
