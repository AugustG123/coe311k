```python
import numpy as np
import time
import matplotlib.pyplot as plt

def naive_matrix_multiplication(A, B):
    n = len(A)
    C = [[0] * n for _ in range(n)]
    for i in range(n):
        for j in range(n):
            for k in range(n):
                C[i][j] += A[i][k] * B[k][j]
                
    return C

timeNaive = []
timeNP = []
for i in range(8):
    A = np.random.rand(3*2^i,3*2^i)
    B = np.random.rand(3*2^i,3*2^i)
    start = time.perf_counter()
    A @ B
    end = time.perf_counter()
    timeNP.append(end-start)
    start = time.perf_counter()
    naive_matrix_multiplication(A,B)
    end = time.perf_counter()
    timeNaive.append(end - start)
    
plt.plot(range(8), timeNaive, label = "Naive")
plt.plot(range(8), timeNP, label = "Numpy")
plt.legend()
plt.show()
```
